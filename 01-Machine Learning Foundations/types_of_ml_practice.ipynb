**Your answer:**
- ML type: **Supervised — Classification**
- Why: every past email already has the correct answer attached (`is_spam`: yes/no), and the thing we're predicting is a *category* (spam vs. not-spam), not a number.


```python
# Problem 1 — Email Spam Filter (Supervised - Classification)

# Each past email = FEATURES (sender, subject, word_count) + a LABEL (is_spam).
# This labeled list is exactly what you'd feed a classification model to train on.
past_emails = [
    {"sender": "boss@company.com",   "subject": "Meeting notes",        "word_count": 120, "is_spam": False},
    {"sender": "winner@lucky.biz",   "subject": "YOU WON $1,000,000",   "word_count": 15,  "is_spam": True},
    {"sender": "mom@gmail.com",      "subject": "Dinner on Sunday?",    "word_count": 40,  "is_spam": False},
    {"sender": "deals@spammyshop.co","subject": "FREE MONEY CLICK NOW", "word_count": 8,   "is_spam": True},
]

for email in past_emails:
    # print each past email's features + its known label -> this is TRAINING data
    print(email)
# {'sender': 'boss@company.com', 'subject': 'Meeting notes', 'word_count': 120, 'is_spam': False}
# {'sender': 'winner@lucky.biz', 'subject': 'YOU WON $1,000,000', 'word_count': 15, 'is_spam': True}
# {'sender': 'mom@gmail.com', 'subject': 'Dinner on Sunday?', 'word_count': 40, 'is_spam': False}
# {'sender': 'deals@spammyshop.co', 'subject': 'FREE MONEY CLICK NOW', 'word_count': 8, 'is_spam': True}

# A brand-new incoming email -> NO label yet, this is what we'd want to PREDICT
new_email = {"sender": "unknown@offers.net", "subject": "Claim your FREE prize now!!!", "word_count": 10}

print(new_email)
# {'sender': 'unknown@offers.net', 'subject': 'Claim your FREE prize now!!!', 'word_count': 10}
# -> a trained model would look at this new_email's features and predict is_spam: True/False.
# We haven't learned HOW to train that model yet (later chapters) -- this cell just shows
# the SHAPE of a classification problem: labeled training data -> unlabeled new data to predict on.
```

**Your answer:**
- ML type: **Supervised — Regression**
- Why: every past house has a known label (its actual sale `price`), and price is a *continuous number*, not a category — that's what makes it regression instead of classification.


```python
# Problem 2 — Predicting House Prices (Supervised - Regression)

# Same idea as Problem 1: FEATURES + a LABEL for every past example.
# The difference: the label ("price") is a NUMBER, not a category -> regression, not classification.
past_houses = [
    {"sqft": 1200, "bedrooms": 2, "distance_km": 5.0, "price_lakhs": 55},
    {"sqft": 2400, "bedrooms": 4, "distance_km": 12.0, "price_lakhs": 95},
    {"sqft": 900,  "bedrooms": 1, "distance_km": 2.0, "price_lakhs": 42},
    {"sqft": 1800, "bedrooms": 3, "distance_km": 8.0, "price_lakhs": 78},
]

for house in past_houses:
    # each row = features + the actual number it sold for (the label)
    print(house)
# {'sqft': 1200, 'bedrooms': 2, 'distance_km': 5.0, 'price_lakhs': 55}
# {'sqft': 2400, 'bedrooms': 4, 'distance_km': 12.0, 'price_lakhs': 95}
# {'sqft': 900, 'bedrooms': 1, 'distance_km': 2.0, 'price_lakhs': 42}
# {'sqft': 1800, 'bedrooms': 3, 'distance_km': 8.0, 'price_lakhs': 78}

# A new house, not yet sold -> no price label -> this is what we'd predict a NUMBER for
new_house = {"sqft": 1500, "bedrooms": 3, "distance_km": 6.0}

print(new_house)
# {'sqft': 1500, 'bedrooms': 3, 'distance_km': 6.0}
# -> a trained regression model would output something like price_lakhs: 68.3 (a number, not a
# yes/no category) -- that continuous-number output is exactly what makes this regression.
```

**Your answer:**
- ML type: **Unsupervised — Clustering**
- Why: there's no `customer_type` label anywhere in the data — we only have raw features (order value, order frequency, recency). The goal is to *discover* natural groupings, not predict a known answer.


```python
# Problem 3 — Grouping Customers with No Labels (Unsupervised - Clustering)

# Compare this to Problem 1/2: notice there's NO label column at all here (no "customer_type").
# We only have raw features -- that missing label is exactly why this is unsupervised.
customers = [
    {"avg_order_value": 250,  "orders_per_month": 8, "days_since_last_order": 2},
    {"avg_order_value": 4000, "orders_per_month": 1, "days_since_last_order": 5},
    {"avg_order_value": 300,  "orders_per_month": 7, "days_since_last_order": 3},
    {"avg_order_value": 50,   "orders_per_month": 0, "days_since_last_order": 90},
    {"avg_order_value": 3800, "orders_per_month": 1, "days_since_last_order": 4},
    {"avg_order_value": 40,   "orders_per_month": 0, "days_since_last_order": 120},
]

for c in customers:
    # just features, no "correct group" attached anywhere
    print(c)
# {'avg_order_value': 250, 'orders_per_month': 8, 'days_since_last_order': 2}
# {'avg_order_value': 4000, 'orders_per_month': 1, 'days_since_last_order': 5}
# {'avg_order_value': 300, 'orders_per_month': 7, 'days_since_last_order': 3}
# {'avg_order_value': 50, 'orders_per_month': 0, 'days_since_last_order': 90}
# {'avg_order_value': 3800, 'orders_per_month': 1, 'days_since_last_order': 4}
# {'avg_order_value': 40, 'orders_per_month': 0, 'days_since_last_order': 120}

# Just by eyeballing the numbers above (before any real clustering algorithm), you can
# already sense 3 rough groups -- this is the human intuition a clustering algorithm
# (like k-means, covered in a later chapter) automates at scale:
#   - frequent small spenders   (rows 1 and 3: low value, many orders, recent)
#   - rare big spenders         (rows 2 and 5: high value, few orders, recent)
#   - inactive / at-risk        (rows 4 and 6: low value, zero orders, long time since last order)
print("Eyeballed groups: frequent small spenders, rare big spenders, inactive/at-risk")
# Eyeballed groups: frequent small spenders, rare big spenders, inactive/at-risk
```

**Your answer:**
- ML type: **Unsupervised — Dimensionality Reduction**
- Why: again no label to predict — the goal is to compress/simplify columns while keeping the useful signal, not predict a category or number for each product.


```python
# Problem 4 — Too Many Columns (Unsupervised - Dimensionality Reduction)

# No real dataset needed here -- just showing the SHAPE of the problem: many columns in,
# few (information-rich) columns out. No label involved anywhere, same as Problem 3.
original_columns = [
    "length_cm", "length_inches", "width_cm", "width_inches",
    "height_cm", "height_inches", "weight_kg", "weight_lbs",
    "color", "material", "price", "category",
    # ... imagine ~290 more columns, many just unit-duplicates of each other
]

print(f"Columns before: {len(original_columns)}")
# Columns before: 12   (in real life this scenario is ~300)

# Dimensionality reduction would compress this down to fewer NEW columns that still
# capture most of the same information (e.g. combining length/width/height into one
# "size" signal, since cm and inches versions carry duplicate information anyway).
reduced_columns = ["size_signal", "weight_signal", "color", "material", "price", "category"]

print(f"Columns after: {len(reduced_columns)}")
# Columns after: 6

# Why unsupervised: there is no "correct answer" column being predicted here at all --
# we're not classifying or predicting a number for each product, we're simplifying the
# FEATURES themselves. That absence of a label is the same signal as Problem 3.
print("No label column involved -> unsupervised, same reasoning as Problem 3")
# No label column involved -> unsupervised, same reasoning as Problem 3
```

**Your answer:**
- ML type: **Reinforcement Learning**
- Why: there's no upfront dataset of "correct routes" (no labels at all). The robot only gets a reward signal (trip time) *after* acting, and must improve its strategy through repeated trial and error.


```python
# Problem 5 — A Delivery Robot Learning a Route (Reinforcement Learning)

# No clean "dataset" here like the earlier problems -- instead we define the 4 RL
# ingredients from theory.md Section 5 directly, since that's what actually describes
# this kind of problem (agent, environment, action, reward), not a labeled table.

agent = "warehouse robot"
environment = "warehouse floor layout (dock, aisles, shelves)"
action = "turn left at aisle 3"          # one example of a choice the agent can make
reward = -12                              # example: took 12 seconds LONGER than the best-known route so far (a penalty)
# (a positive reward, e.g. +5, would mean 5 seconds FASTER than expected -- good outcome)

print(f"Agent: {agent}")
# Agent: warehouse robot
print(f"Environment: {environment}")
# Environment: warehouse floor layout (dock, aisles, shelves)
print(f"Example action: {action}")
# Example action: turn left at aisle 3
print(f"Reward received after this action: {reward}")
# Reward received after this action: -12

# Unlike Problems 1-4, there's no upfront labeled/unlabeled dataset to point at --
# the "data" here is generated LIVE, attempt by attempt, as the robot acts and receives
# a reward/penalty. Over thousands of these attempts, it adjusts its policy (its
# strategy of "in this situation, take this action") to chase a higher total reward.
policy_so_far = {"at aisle 3": "turn left"}   # the robot's current learned strategy, so far
print(policy_so_far)
# {'at aisle 3': 'turn left'}
```

**Your answer — two separate ML problems here:**
- Loan repay/default prediction: **Supervised — Classification** — every past loan has a known outcome (`repaid`/`defaulted`), and that outcome is a category.
- The earlier customer grouping step: **Unsupervised — Clustering** — deliberately done *without* using the repaid/defaulted label, purely grouping by spending behavior to explore the data first.


```python
# Problem 6 — The Tricky One: TWO separate ML problems in one scenario

# --- Problem 6a: Loan repay/default prediction -> Supervised - Classification ---
# Every past applicant has a known outcome (repaid or defaulted) -- a category label.
past_loan_applications = [
    {"income": 45000, "credit_score": 720, "loan_amount": 200000, "outcome": "repaid"},
    {"income": 22000, "credit_score": 580, "loan_amount": 150000, "outcome": "defaulted"},
    {"income": 60000, "credit_score": 750, "loan_amount": 300000, "outcome": "repaid"},
    {"income": 18000, "credit_score": 540, "loan_amount": 100000, "outcome": "defaulted"},
]

for app in past_loan_applications:
    # features + the known label ("outcome") for each past applicant
    print(app)
# {'income': 45000, 'credit_score': 720, 'loan_amount': 200000, 'outcome': 'repaid'}
# {'income': 22000, 'credit_score': 580, 'loan_amount': 150000, 'outcome': 'defaulted'}
# {'income': 60000, 'credit_score': 750, 'loan_amount': 300000, 'outcome': 'repaid'}
# {'income': 18000, 'credit_score': 540, 'loan_amount': 100000, 'outcome': 'defaulted'}

new_applicant = {"income": 50000, "credit_score": 690, "loan_amount": 250000}
print(new_applicant)
# {'income': 50000, 'credit_score': 690, 'loan_amount': 250000}
# -> predict outcome: repaid/defaulted (a category) for this new applicant

print("---")
# ---

# --- Problem 6b: Earlier customer grouping step -> Unsupervised - Clustering ---
# SAME underlying customers, but this step deliberately does NOT use the outcome label --
# only spending behavior, to explore natural groupings before the loan model is even built.
customer_spending_behavior = [
    {"income": 45000, "avg_monthly_spend": 12000},
    {"income": 22000, "avg_monthly_spend": 18000},   # spends more than they earn -> risk signal, found without using "outcome" at all
    {"income": 60000, "avg_monthly_spend": 15000},
    {"income": 18000, "avg_monthly_spend": 17000},
]

for c in customer_spending_behavior:
    # features only -- notice "outcome" is intentionally absent from this view of the data
    print(c)
# {'income': 45000, 'avg_monthly_spend': 12000}
# {'income': 22000, 'avg_monthly_spend': 18000}
# {'income': 60000, 'avg_monthly_spend': 15000}
# {'income': 18000, 'avg_monthly_spend': 17000}

# Same customers, two different problems: one WITH a label in play (6a, supervised),
# one WITHOUT (6b, unsupervised) -- the presence/absence of the label is what decides it,
# not the customers or the columns being similar.
```
